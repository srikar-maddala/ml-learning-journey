```python
from datasets import load_dataset
dataset = load_dataset("stanfordnlp/imdb")
```

```python
print(dataset)
print(dataset['train'][0])
dataset.shape
```

```python
#Independent features
train_dataset=dataset["train"]
X_train=list(train_dataset["text"])
Y_train=list(train_dataset["label"])
```

```python
test_dataset=dataset["test"]
X_test=list(test_dataset["text"])
Y_test=list(test_dataset["label"])
```

```python
from transformers import DistilBertTokenizerFast
tokenizer=DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')
```

```python
train_encoding=tokenizer(X_train,truncation=True,padding=True,max_length=256)
test_encoding=tokenizer(X_test,truncation=True,padding=True,max_length=256)
```

```python
train_encoding['input_ids'][0]
```

```python
#Convert these encoding into a datset objects
train_dataset
```

```python
from datasets import Dataset
```

```python
train_dataset = Dataset.from_dict({
    'input_ids': train_encoding['input_ids'],
    'attention_mask': train_encoding['attention_mask'],
    'labels': Y_train
})
```

```python
test_dataset = Dataset.from_dict({
    'input_ids': test_encoding['input_ids'],
    'attention_mask': test_encoding['attention_mask'],
    'labels': Y_test
})
```

```python
train_dataset[0]
```

```python
train_dataset.set_format(
    "torch",
    columns=[
        "input_ids",
        "attention_mask",
        "labels"
    ]
)
```

```python
test_dataset.set_format(
    "torch",
    columns=[
        "input_ids",
        "attention_mask",
        "labels"
    ]
)
```

```python
train_dataset[0]
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=16
)
```

```python
from transformers import DistilBertForSequenceClassification

model = DistilBertForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2
)
```

```python
from transformers import DistilBertForSequenceClassification,Trainer,TrainingArguments
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=3,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    learning_rate=2e-5,
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_dir="./logs"
)
```

```python
trainer =Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset

)
```

```python
print(model)
```

```python
trainer.train()
```

```python
trainer.evaluate()
```

```python
trainer.save_model("./distilbert-imdb-model")
tokenizer.save_pretrained("./distilbert-imdb-model")
```

```python
text = "This movie was fantastic. The acting was amazing!"
```

```python
inputs = tokenizer(
    text,
    return_tensors="pt",
    truncation=True,
    padding=True
)
```

```python
inputs = {key:value.to(model.device) for key,value in inputs.items()}
```

```python
with torch.no_grad():

    outputs = model(**inputs)

logits = outputs.logits

prediction = torch.argmax(logits, dim=1)

print(prediction)
```
